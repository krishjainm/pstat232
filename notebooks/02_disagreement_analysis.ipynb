# 02 — Disagreement Analysis

This notebook analyzes the agreement/disagreement distribution and explores what disagreement cases look like.

```python
import sys
sys.path.insert(0, "..")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")

train = pd.read_parquet("../data/processed/train.parquet")
test = pd.read_parquet("../data/processed/test.parquet")

print(f"Train: {len(train)}, Test: {len(test)}")
print(f"\nDisagreement columns: {[c for c in train.columns if 'disagree' in c or 'agreement' in c]}")
```

```python
# Disagreement distribution
print("Training set disagreement distribution:")
print(train["disagreement_group"].value_counts())
print(f"\nDisagreement rate: {(train['agreement_status'] == 'disagreement').mean():.3f}")
```

```python
# Example disagreement cases
disagree = train[train["agreement_status"] == "disagreement"]
print(f"Total disagreement cases in train: {len(disagree)}\n")

for _, row in disagree.head(5).iterrows():
    print(f"Rating: {row['rating']} | Label: {row['label']} | Text sentiment: {row['text_sentiment_label']} (conf: {row['text_sentiment_confidence']:.2f})")
    print(f"  Group: {row['disagreement_group']}")
    print(f"  Text: {row['review_text'][:150]}...\n")
```

```python
# Text sentiment confidence distribution by agreement status
fig, ax = plt.subplots(figsize=(8, 4))
for status in ["agreement", "disagreement"]:
    subset = train[train["agreement_status"] == status]
    ax.hist(subset["text_sentiment_confidence"], bins=30, alpha=0.6, label=status)
ax.set_xlabel("Text Sentiment Confidence")
ax.set_ylabel("Count")
ax.set_title("Text Sentiment Confidence by Agreement Status")
ax.legend()
plt.show()
```
