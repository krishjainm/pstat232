# 04 — Figures for Report

This notebook displays all generated figures. Run `scripts/08_generate_figures.py` first.

```python
from IPython.display import Image, display
import os

fig_dir = "../reports/figures"
figures = [
    "class_distribution.png",
    "agreement_distribution.png",
    "model_accuracy_by_group.png",
    "f1_by_group.png",
    "calibration_curves.png",
    "confidence_correct_vs_incorrect.png",
    "modality_dominance.png",
    "high_confidence_errors.png",
    "confusion_matrices.png",
    "disagreement_severity_trend.png",
]

for fig_name in figures:
    path = os.path.join(fig_dir, fig_name)
    if os.path.exists(path):
        print(f"\n--- {fig_name} ---")
        display(Image(filename=path, width=600))
    else:
        print(f"  [Not found: {fig_name}]")
```
