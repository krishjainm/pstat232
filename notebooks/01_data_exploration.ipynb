# 01 — Data Exploration

This notebook explores the Amazon Reviews 2023 dataset (All_Beauty category) after preprocessing.

Key questions:
- How are ratings distributed?
- What does the text look like?
- What metadata features are available?
- What is the class balance after binarization?

```python
import sys
sys.path.insert(0, "..")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")

train = pd.read_parquet("../data/processed/train.parquet")
val = pd.read_parquet("../data/processed/val.parquet")
test = pd.read_parquet("../data/processed/test.parquet")

print(f"Train: {len(train)}, Val: {len(val)}, Test: {len(test)}")
train.head()
```

```python
# Class distribution
fig, ax = plt.subplots(figsize=(5, 3))
train["label"].value_counts().sort_index().plot(kind="bar", ax=ax, color=["salmon", "skyblue"])
ax.set_xticklabels(["Negative (0)", "Positive (1)"], rotation=0)
ax.set_title("Training Set Class Distribution")
plt.show()
```

```python
# Review length distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
train["review_length_words"].hist(bins=50, ax=axes[0])
axes[0].set_title("Review Length (words)")
axes[0].set_xlabel("Words")

train["review_length_chars"].hist(bins=50, ax=axes[1])
axes[1].set_title("Review Length (chars)")
axes[1].set_xlabel("Characters")
plt.tight_layout()
plt.show()
```

```python
# Metadata feature summary
meta_cols = ["review_length_words", "helpful_vote", "verified_purchase",
             "product_average_rating", "product_rating_number", "product_price", "review_year"]
available = [c for c in meta_cols if c in train.columns]
train[available].describe()
```

```python
# Sample reviews
print("=== Sample Positive Reviews ===")
for t in train[train["label"] == 1]["review_text"].head(3):
    print(f"  {t[:200]}...\n")

print("=== Sample Negative Reviews ===")
for t in train[train["label"] == 0]["review_text"].head(3):
    print(f"  {t[:200]}...\n")
```
