# 03 — Model Results

This notebook presents the evaluation results: overall metrics, group metrics, calibration, and modality dominance.

```python
import sys
sys.path.insert(0, "..")

import pandas as pd
import numpy as np

main_results = pd.read_csv("../reports/tables/main_results.csv")
group_results = pd.read_csv("../reports/tables/group_results.csv")
cal_results = pd.read_csv("../reports/tables/calibration_results.csv")
dom_results = pd.read_csv("../reports/tables/modality_dominance_results.csv")
```

```python
print("=== Main Results ===")
display(main_results.round(4))
```

```python
print("=== Group Results ===")
pivot = group_results.pivot_table(index="group", columns="model", values="accuracy").round(4)
display(pivot)
```

```python
print("=== Calibration Results ===")
display(cal_results.round(4))
```

```python
print("=== Modality Dominance ===")
display(dom_results)
```
